## scrape match statistics

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import re

# We'll use ScraperFC's browser to get the page (bypasses Cloudflare)
import ScraperFC as sfc
fb = sfc.FBref()

url = "https://fbref.com/en/matches/a68e623d/Monaco-Toulouse-August-4-2017-Ligue-1"
soup = fb._get_soup(url)

# Extract team names from the first row
team_stats_table = soup.select_one('#team_stats > table > tbody')
rows = team_stats_table.find_all('tr')

row_data = {}

# First row: team names
team_cells = rows[0].find_all('th')
row_data['home_team'] = team_cells[0].get_text(strip=True)
row_data['away_team'] = team_cells[1].get_text(strip=True)

# Process stat rows (they come in pairs: header row + data row)
i = 1
while i < len(rows):
    # Check if this is a header row (contains <th colspan="2">)
    header = rows[i].find('th')
    if header and header.get('colspan') == '2':
        stat_name = header.get_text(strip=True).lower().replace(' ', '_')
        data_row = rows[i + 1]
        cells = data_row.find_all('td')
        
        if stat_name == 'cards':
            # Cards: count yellow_card spans
            home_cards = len(cells[0].find_all('span', class_='yellow_card'))
            away_cards = len(cells[1].find_all('span', class_='yellow_card'))
            row_data[f'home_{stat_name}'] = home_cards
            row_data[f'away_{stat_name}'] = away_cards
        else:
            # Extract percentage values
            home_text = cells[0].get_text(strip=True)
            away_text = cells[1].get_text(strip=True)
            
            # Extract the main percentage
            home_pct = re.findall(r'(\d+)%', home_text)
            away_pct = re.findall(r'(\d+)%', away_text)
            
            row_data[f'home_{stat_name}'] = home_pct[0] if home_pct else None
            row_data[f'away_{stat_name}'] = away_pct[0] if away_pct else None
        
        i += 2
    else:
        i += 1

# Create DataFrame
df = pd.DataFrame([row_data])
print(df.to_string())
print(f"\nColumns: {list(df.columns)}")
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import re

import ScraperFC as sfc
fb = sfc.FBref()

url = "https://fbref.com/en/matches/a68e623d/Monaco-Toulouse-August-4-2017-Ligue-1"
soup = fb._get_soup(url)

# Extract team_stats_extra
team_stats_extra = soup.select_one('#team_stats_extra')

row_data = {}

# Each column is a <div> inside #team_stats_extra
columns = team_stats_extra.find_all('div', recursive=False)

for col_idx, column in enumerate(columns):
    # Get all divs inside this column
    divs = column.find_all('div')
    
    # The first 3 divs are headers: home_team, empty, away_team
    home_team = divs[0].get_text(strip=True)
    away_team = divs[2].get_text(strip=True)
    
    # The rest come in groups of 3: home_value, stat_name, away_value
    for i in range(3, len(divs), 3):
        if i + 2 < len(divs):
            home_val = divs[i].get_text(strip=True)
            stat_name = divs[i + 1].get_text(strip=True).lower().replace(' ', '_')
            away_val = divs[i + 2].get_text(strip=True)
            
            row_data[f'home_{stat_name}'] = home_val
            row_data[f'away_{stat_name}'] = away_val

# Create DataFrame
df = pd.DataFrame([row_data])
print(df.to_string(max_colwidth=20))
print(f"\nColumns: {list(df.columns)}")
print(f"Total stats: {len(df.columns)}")
```

## scrape HOME player aggregate statistics (shots, penalties...)

```python
import ScraperFC as sfc
from bs4 import BeautifulSoup
import pandas as pd
import re

fb = sfc.FBref()
url = "https://fbref.com/en/matches/a68e623d/Monaco-Toulouse-August-4-2017-Ligue-1"
soup = fb._get_soup(url)

# Find home team ID from the all_player_stats div
home_div = soup.find('div', id=re.compile(r'all_player_stats_'))
home_id = home_div['id'].replace('all_player_stats_', '') if home_div else None

print(f"Home team ID: {home_id}")

# Extract the tfoot row (aggregated team totals)
row_data = {}
if home_id:
    tfoot = soup.select_one(f'#stats_{home_id}_summary tfoot tr')
    if tfoot:
        for td in tfoot.find_all(['th', 'td']):
            stat = td.get('data-stat', '')
            if stat and stat != 'player':
                val = td.get_text(strip=True)
                row_data[f'Home_{stat}'] = val

# Display results
print(f"\nExtracted {len(row_data)} stats:")
for key, value in row_data.items():
    print(f"  {key}: {value}")
```

## scrape HOME goalkeeper stats

```python
import ScraperFC as sfc
from bs4 import BeautifulSoup
import pandas as pd
import re

fb = sfc.FBref()
url = "https://fbref.com/en/matches/a68e623d/Monaco-Toulouse-August-4-2017-Ligue-1"
soup = fb._get_soup(url)

# Find home team ID
home_div = soup.find('div', id=re.compile(r'all_player_stats_'))
home_id = home_div['id'].replace('all_player_stats_', '') if home_div else None

print(f"Home team ID: {home_id}")

# Extract goalkeeper stats
row_data = {}
if home_id:
    keeper_table = soup.select_one(f'#keeper_stats_{home_id} tbody')
    if keeper_table:
        # Get the first (and usually only) keeper row
        for row in keeper_table.find_all('tr'):
            for td in row.find_all(['th', 'td']):
                stat = td.get('data-stat', '')
                if stat and stat != 'player':
                    val = td.get_text(strip=True)
                    row_data[f'Home_keeper_{stat}'] = val

# Display results
print(f"\nExtracted {len(row_data)} keeper stats:")
for key, value in row_data.items():
    print(f"  {key}: {value}")
```

```python
import ScraperFC as sfc
from bs4 import BeautifulSoup
import pandas as pd
import re

fb = sfc.FBref()
url = "https://fbref.com/en/matches/a68e623d/Monaco-Toulouse-August-4-2017-Ligue-1"
soup = fb._get_soup(url)

# Get match info
league = "Ligue 1"  # Could extract from soup
season = "2017-2018"  # Could extract from soup
game_id = url.split('/')[-2]

# Find team IDs
home_div = soup.find('div', id=re.compile(r'all_player_stats_'))
home_id = home_div['id'].replace('all_player_stats_', '') if home_div else None

# Get team names from the match stats section
team_stats = soup.select_one('#team_stats > table > tbody tr')
home_team = team_stats.find_all('th')[0].get_text(strip=True) if team_stats else None

print(f"Game ID: {game_id}")
print(f"League: {league}")
print(f"Season: {season}")
print(f"Home team: {home_team}")

# Extract player rows from <tbody> (NOT <tfoot>)
players = []
if home_id:
    tbody = soup.select_one(f'#stats_{home_id}_summary tbody')
    if tbody:
        for row in tbody.find_all('tr'):
            player_data = {}
            
            # Get player name
            player_th = row.find('th', {'data-stat': 'player'})
            if player_th:
                player_data['player'] = player_th.get_text(strip=True)
            
            # Extract all td elements
            for td in row.find_all('td'):
                stat = td.get('data-stat', '')
                val = td.get_text(strip=True)
                
                if stat == 'shirtnumber':
                    player_data['jersey_number'] = val
                elif stat == 'position':
                    player_data['position'] = val
                elif stat == 'minutes':
                    player_data['minutes_played'] = val
            
            if player_data.get('player'):
                player_data['league'] = league
                player_data['season'] = season
                player_data['game'] = game_id
                player_data['team'] = home_team
                # Starter if minutes > 0 and not subbed in late (rough heuristic)
                # Actually: if they have a shirt number and minutes > ~45
                mins = int(player_data.get('minutes_played', 0))
                player_data['is_starter'] = mins >= 45
                
                players.append(player_data)

# Create DataFrame
df_players = pd.DataFrame(players)
print(f"\nExtracted {len(df_players)} players:")
print(df_players.to_string())
```

```python
import ScraperFC as sfc
from bs4 import BeautifulSoup
import pandas as pd
import re

fb = sfc.FBref()
url = "https://fbref.com/en/matches/a68e623d/Monaco-Toulouse-August-4-2017-Ligue-1"
soup = fb._get_soup(url)

# Find both team IDs from all_player_stats divs
team_ids = []
for div in soup.find_all('div', id=re.compile(r'all_player_stats_')):
    team_id = div['id'].replace('all_player_stats_', '')
    team_ids.append(team_id)

home_id = team_ids[0] if len(team_ids) > 0 else None
away_id = team_ids[1] if len(team_ids) > 1 else None

print(f"Home team ID: {home_id}")
print(f"Away team ID: {away_id}")

# ============================================================
# 1. Player summary totals (<tfoot>) - Home & Away
# ============================================================
player_totals = {}

for side, team_id in [('Home', home_id), ('Away', away_id)]:
    if team_id:
        tfoot = soup.select_one(f'#stats_{team_id}_summary tfoot tr')
        if tfoot:
            for td in tfoot.find_all(['th', 'td']):
                stat = td.get('data-stat', '')
                if stat and stat != 'player':
                    val = td.get_text(strip=True)
                    player_totals[f'{side}_{stat}'] = val

# ============================================================
# 2. Per-player lineup info - Home & Away
# ============================================================
players = []

for side, team_id in [('Home', home_id), ('Away', away_id)]:
    if team_id:
        # Get team name from the caption
        caption = soup.select_one(f'#stats_{team_id}_summary caption')
        team_name = caption.get_text(strip=True).replace(' Player Stats Table', '') if caption else team_id
        
        tbody = soup.select_one(f'#stats_{team_id}_summary tbody')
        if tbody:
            for row in tbody.find_all('tr'):
                player_data = {'side': side, 'team': team_name, 'team_id': team_id}
                
                # Player name
                player_th = row.find('th', {'data-stat': 'player'})
                if player_th:
                    player_data['player'] = player_th.get_text(strip=True)
                
                # Other stats
                for td in row.find_all('td'):
                    stat = td.get('data-stat', '')
                    val = td.get_text(strip=True)
                    if stat == 'shirtnumber':
                        player_data['jersey_number'] = val
                    elif stat == 'position':
                        player_data['position'] = val
                    elif stat == 'minutes':
                        player_data['minutes_played'] = val
                
                if player_data.get('player'):
                    mins = int(player_data.get('minutes_played', 0))
                    player_data['is_starter'] = mins >= 45
                    players.append(player_data)

# ============================================================
# 3. Keeper stats - Home & Away
# ============================================================
keeper_stats = {}

for side, team_id in [('Home', home_id), ('Away', away_id)]:
    if team_id:
        keeper_table = soup.select_one(f'#keeper_stats_{team_id} tbody')
        if keeper_table:
            for row in keeper_table.find_all('tr'):
                for td in row.find_all(['th', 'td']):
                    stat = td.get('data-stat', '')
                    if stat and stat != 'player':
                        val = td.get_text(strip=True)
                        keeper_stats[f'{side}_keeper_{stat}'] = val

# ============================================================
# Display results
# ============================================================
print(f"\n--- Player Totals ({len(player_totals)} stats) ---")
for k, v in player_totals.items():
    print(f"  {k}: {v}")

print(f"\n--- Players ({len(players)} players) ---")
df_players = pd.DataFrame(players)
print(df_players[['side', 'player', 'jersey_number', 'position', 'minutes_played', 'is_starter']].to_string())

print(f"\n--- Keeper Stats ({len(keeper_stats)} stats) ---")
for k, v in keeper_stats.items():
    print(f"  {k}: {v}")
```

## test full scraping on one match

```python
import ScraperFC as sfc
from bs4 import BeautifulSoup
import pandas as pd
import re

fb = sfc.FBref()

def extract_match_data(url: str) -> dict:
    """
    Extract all match-level data from a single FBref match page.
    Returns a dictionary with one row of aggregated features.
    """
    soup = fb._get_soup(url)
    
    row = {}
    
    # Extract match ID from URL
    match_id = url.split('/')[-2]
    row['match_id'] = match_id
    row['url'] = url
    
    # ============================================================
    # Find team IDs and names
    # ============================================================
    team_ids = []
    for div in soup.find_all('div', id=re.compile(r'all_player_stats_')):
        team_ids.append(div['id'].replace('all_player_stats_', ''))
    
    home_id = team_ids[0] if len(team_ids) > 0 else None
    away_id = team_ids[1] if len(team_ids) > 1 else None
    
    # Get team names from match stats section
    team_stats_tbody = soup.select_one('#team_stats > table > tbody')
    if team_stats_tbody:
        first_row = team_stats_tbody.find_all('tr')[0]
        team_cells = first_row.find_all('th')
        home_team = team_cells[0].get_text(strip=True) if len(team_cells) > 0 else None
        away_team = team_cells[1].get_text(strip=True) if len(team_cells) > 1 else None
        row['home_team'] = home_team
        row['away_team'] = away_team
    
    # ============================================================
    # 1. Match stats from #team_stats
    # ============================================================
    if team_stats_tbody:
        rows = team_stats_tbody.find_all('tr')
        i = 1
        while i < len(rows):
            header = rows[i].find('th')
            if header and header.get('colspan') == '2':
                stat_name = header.get_text(strip=True).lower().replace(' ', '_')
                data_row = rows[i + 1]
                cells = data_row.find_all('td')
                
                if stat_name == 'cards':
                    home_val = len(cells[0].find_all('span', class_='yellow_card'))
                    away_val = len(cells[1].find_all('span', class_='yellow_card'))
                else:
                    home_text = cells[0].get_text(strip=True)
                    away_text = cells[1].get_text(strip=True)
                    home_pct = re.findall(r'(\d+)%', home_text)
                    away_pct = re.findall(r'(\d+)%', away_text)
                    home_val = home_pct[0] if home_pct else None
                    away_val = away_pct[0] if away_pct else None
                
                row[f'Home_{stat_name}'] = home_val
                row[f'Away_{stat_name}'] = away_val
                i += 2
            else:
                i += 1
    
    # ============================================================
    # 2. Match extras from #team_stats_extra
    # ============================================================
    team_stats_extra = soup.select_one('#team_stats_extra')
    if team_stats_extra:
        columns = team_stats_extra.find_all('div', recursive=False)
        for column in columns:
            divs = column.find_all('div')
            for i in range(3, len(divs), 3):
                if i + 2 < len(divs):
                    home_val = divs[i].get_text(strip=True)
                    stat_name = divs[i + 1].get_text(strip=True).lower().replace(' ', '_')
                    away_val = divs[i + 2].get_text(strip=True)
                    row[f'Home_{stat_name}'] = home_val
                    row[f'Away_{stat_name}'] = away_val
    
    # ============================================================
    # 3. Player summary totals from <tfoot>
    # ============================================================
    for side, team_id in [('Home', home_id), ('Away', away_id)]:
        if team_id:
            tfoot = soup.select_one(f'#stats_{team_id}_summary tfoot tr')
            if tfoot:
                for td in tfoot.find_all(['th', 'td']):
                    stat = td.get('data-stat', '')
                    if stat and stat != 'player':
                        val = td.get_text(strip=True)
                        row[f'{side}_{stat}'] = val
    
    # ============================================================
    # 4. Keeper stats
    # ============================================================
    for side, team_id in [('Home', home_id), ('Away', away_id)]:
        if team_id:
            keeper_table = soup.select_one(f'#keeper_stats_{team_id} tbody')
            if keeper_table:
                for keeper_row in keeper_table.find_all('tr'):
                    for td in keeper_row.find_all(['th', 'td']):
                        stat = td.get('data-stat', '')
                        if stat and stat != 'player':
                            val = td.get_text(strip=True)
                            row[f'{side}_keeper_{stat}'] = val
    
    return row


# ============================================================
# Test with a single match
# ============================================================
url = "https://fbref.com/en/matches/a68e623d/Monaco-Toulouse-August-4-2017-Ligue-1"
row = extract_match_data(url)

df = pd.DataFrame([row])
print(f"Shape: {df.shape[0]} rows × {df.shape[1]} columns")
print(f"\nAll columns:")
for i, col in enumerate(df.columns):
    print(f"  {i:3d}: {col} = {df[col].iloc[0]}")
```
