```python
# get the unique team names from data/lineups_values_ratings_games.csv (column name: Home Team)
import pandas as pd
df = pd.read_csv("data/lineups_values_ratings_games.csv")
teams = df["Home Team"].unique()
print(teams)

```

```python
df2 = pd.read_csv("data/big5_xg_clean.csv")
teams2 = df2["home_team_mapped"].unique()
print(teams2)
```

```python
# analyze the differences between teams and teams2, create a mapping between the two sets of team names
mapping = {}
for team in teams:
    if team in teams2:
        mapping[team] = team
    else:
        print(f"Team {team} is not in teams2")  
        
```
